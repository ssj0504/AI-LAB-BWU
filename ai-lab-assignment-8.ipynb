```python
#WAP in python program to implement the Minmax algorithm for a two-player game

#1. Implement the Minimax algorithm using recursion.
#2. Consider MAX and MIN players.
#3. Use the following leaf-node values:

   #3, 5, 2, 9, 12, 5, 23, 23

#4. Determine the optimal score for the MAX player.
#5. Display the final optimal score as output.



import math

def minimax(depth, nodeIndex, isMax, scores, max_depth):
    # Base case: leaf node reached
    if depth == max_depth:
        return scores[nodeIndex]

    # If current move is MAX player, find the maximum attainable value
    if isMax:
        return max(
            minimax(depth + 1, nodeIndex * 2, False, scores, max_depth),
            minimax(depth + 1, nodeIndex * 2 + 1, False, scores, max_depth)
        )
    # If current move is MIN player, find the minimum attainable value
    else:
        return min(
            minimax(depth + 1, nodeIndex * 2, True, scores, max_depth),
            minimax(depth + 1, nodeIndex * 2 + 1, True, scores, max_depth)
        )

# Leaf-node values
scores = [3, 5, 2, 9, 12, 5, 23, 23]

# Calculate the maximum depth of the tree based on the number of leaves
max_depth = math.log2(len(scores))

# Determine the optimal score for the MAX player starting at the root (depth 0, index 0, isMax=True)
optimal_score = minimax(0, 0, True, scores, max_depth)

#Display Results
print(f"Leaf Node Values: {scores}")
print("The optimal score for the MAX player is:", optimal_score)
```

```python
#WAP in python to implement the Alpha-Beta Pruning algorithm for a two-player game,Use the leaf-node values 3,5,2,9,12,5,23,23.
#Determine the optimal score for the MAX player and display which branches are pruned during the search.



import math

# Define initial values for alpha and beta
MAX, MIN = math.inf, -math.inf

def alphabeta(depth, nodeIndex, isMax, scores, alpha, beta, max_depth):
    # Base case: leaf node reached
    if depth == max_depth:
        return scores[nodeIndex]
 
    if isMax:
        best = MIN
        # Explore left and right children (2 branches per node)
        for i in range(2):
            childIndex = nodeIndex * 2 + i
            val = alphabeta(depth + 1, childIndex, False, scores, alpha, beta, max_depth)
            best = max(best, val)
            alpha = max(alpha, best)
            
            # Alpha-Beta Pruning condition
            if beta <= alpha:
                # If we prune after evaluating the first child (i=0), the second child is pruned
                if i == 0: 
                    pruned_index = childIndex + 1
                    if depth + 1 == max_depth:
                        print(f"Pruned leaf node at index {pruned_index} (value: {scores[pruned_index]}) under MAX node (Depth: {depth}, Index: {nodeIndex})")
                    else:
                        print(f"Pruned right subtree under MAX node (Depth: {depth}, Index: {nodeIndex})")
                break
        return best
        
    else:
        best = MAX
        # Explore left and right children
        for i in range(2):
            childIndex = nodeIndex * 2 + i
            val = alphabeta(depth + 1, childIndex, True, scores, alpha, beta, max_depth)
            best = min(best, val)
            beta = min(beta, best)
            
            # Alpha-Beta Pruning condition
            if beta <= alpha:
                if i == 0:
                    pruned_index = childIndex + 1
                    if depth + 1 == max_depth:
                        print(f"Pruned leaf node at index {pruned_index} (value: {scores[pruned_index]}) under MIN node (Depth: {depth}, Index: {nodeIndex})")
                    else:
                        print(f"Pruned right subtree under MIN node (Depth: {depth}, Index: {nodeIndex})")
                break
        return best

# Leaf-node values provided
scores = [3, 5, 2, 9, 12, 5, 23, 23]

# Calculate the maximum depth of the tree
max_depth = int(math.log2(len(scores)))

print("--- Alpha-Beta Pruning Execution ---\n")

# Determine the optimal score starting at the root (depth 0, index 0, isMax=True)
optimal_score = alphabeta(0, 0, True, scores, MIN, MAX, max_depth)

print("\n--- Final Results ---")
print("Leaf Node Values:", scores)
print("The optimal score for the MAX player is:", optimal_score)
```
