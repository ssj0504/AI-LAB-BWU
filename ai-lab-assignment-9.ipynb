```python
print("STUDENT PERFORMANCE EVALUATION SYSTEM")
print("-" * 37)

# Taking inputs from the user
attendance = float(input("Enter attendance percentage: "))
marks = float(input("Enter marks percentage: "))
assignment = float(input("Enter assignment percentage: "))
internal = float(input("Enter internal assessment percentage: "))

print("\nEvaluating Performance...")

# IF-THEN Rules for Classification
if attendance >= 90 and marks >= 85 and assignment >= 80 and internal >= 80:
    performance = "Excellent"
elif attendance >= 75 and marks >= 65 and assignment >= 65 and internal >= 65:
    performance = "Good"
elif attendance >= 60 and marks >= 50 and assignment >= 50 and internal >= 50:
    performance = "Average"
else:
    performance = "Poor"

# Displaying the final result
print(f"Final Classification: {performance}")
```

```python
print("MEDICAL DIAGNOSIS EXPERT SYSTEM")
print("-" * 35)

# Collect symptom inputs from the user
print("Please answer with 'yes' or 'no' for the following symptoms:")
fever = input("Do you have a fever? ").strip().lower() == 'yes'
cough = input("Do you have a cough? ").strip().lower() == 'yes'
headache = input("Do you have a headache? ").strip().lower() == 'yes'
sore_throat = input("Do you have a sore throat? ").strip().lower() == 'yes'

print("\nAnalyzing Symptoms...")

# IF-THEN Rules for Medical Diagnosis
if fever and cough and headache and sore_throat:
    diagnosis = "Flu (Influenza) or COVID-19"
elif fever and cough and sore_throat and not headache:
    diagnosis = "Severe Common Cold or Respiratory Infection"
elif cough and sore_throat and not fever and not headache:
    diagnosis = "Common Cold"
elif fever and headache and not cough and not sore_throat:
    diagnosis = "Viral Fever"
elif headache and not fever and not cough and not sore_throat:
    diagnosis = "Tension Headache"
elif sore_throat and not fever and not cough and not headache:
    diagnosis = "Strep Throat or Localized Infection"
elif not fever and not cough and not headache and not sore_throat:
    diagnosis = "No specific illness detected based on these symptoms. You appear healthy."
else:
    diagnosis = "Unidentified condition. Please consult a doctor for a precise diagnosis."

# Displaying the final result
print("-" * 35)
print(f"Possible Diagnosis: {diagnosis}")
```
